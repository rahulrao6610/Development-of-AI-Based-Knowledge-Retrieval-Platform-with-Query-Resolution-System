```python
# Explain instance methods, class methods, and static methods in Python.
# Requirements are Create a class named Employee, Implement an instance method to display employee details, Implement a class method using @classmethod to access or modify a class variable.
# Implement a static method using @staticmethod to perform a simple calculation, Explain the differences between the three method types.

'''
Instance Method:
Defined with def method(self, ...).
Operates on instance variables (unique to each object).
Accessed via object.

Class Method:
Defined with @classmethod.
First parameter is cls (refers to the class itself).
Operates on class variables (shared across all objects).
Accessed via class or object.

Static Method:
Defined with @staticmethod.
Does not take self or cls.
Independent utility function inside the class.
Accessed via class or object.
'''

class emp:
  company_name="tulips"
  def __init__(self,emp_id,name,sal):
    self.emp_id=emp_id
    self.name=name
    self.sal=sal
  def display(self):
    print("Employee ID:",self.emp_id)
    print("Employee Name:",self.name)
    print("Employee Salary:",self.sal)
    print("-" * 30)
  @classmethod
  def changecmpy(cls,new_name):
    cls.company_name=new_name
    print("Comapany name changed to:",cls.company_name)
  @staticmethod
  def cal_bonus(sal,percentage):
    return sal*(percentage/100)
#create emp object
emp1=emp(101,"Rahul",50000)
emp2=emp(102,"prashanth",100000)
emp3=emp(103,"Prathushya",30000)
#instance method
emp1.display()
emp2.display()
emp2.display()
#class method
emp.changecmpy("guntur karam")
# After changing company name
emp1.display()
emp2.display()
emp3.display()

# static method
bonus=emp.cal_bonus(emp2.sal,10)
print("Bonus for emp2:",bonus)
```

```python
# Explain inheritance in Python using a real-world example.
# Requirements are Create a parent class named Person with name and age.
# Create a child class named Student that inherits from Person,Add roll_number and course to Student.
# Use super() to call the parent constructor,Display all student details,Explain code reuse and inheritance.

'''
Inheritance Concept--
Inheritance allows a class (child) to reuse attributes and methods from another class (parent).
This promotes code reuse and avoids duplication.

Example: A Student is a type of Person, so it inherits common attributes like name and age.
'''
# Parent class
class person:
  def __init__(self,name,age):
    self.name=name
    self.age=age
  def display(self):
    print("Name:",self.name)
    print("Age:",self.age)
# Child class inheriting from Person
class student(person):
  def __init__(self,name,age,roll_no,course):
    super().__init__(name,age)
    self.roll_no=roll_no
    self.course=course
 # Call parent constructor using super()
  def display(self):
    # Reuse parent method
    super().display()
    print("Roll No:",self.roll_no)
    print("Course:",self.course)
    print("-"*20)

# Create student objects
s1=student("Rahul",21,101,"Computer Science")
s2=student("Sneha",22,102,"Data Science")

# Display details
s1.display()
s2.display()


```

```python
# Write a Python program demonstrating polymorphism.
# Requirements are Create a base class named Animal with a sound() method,Create Dog, Cat, and Cow classes with their own sound() methods.
# Demonstrate polymorphism by calling sound() on different objects,Explain how the same method name produces different behavior.
# Provide executable Python 3 code and sample output.

'''
Polymorphism Concept---
Polymorphism means many forms.
In Python, it allows the same method name (sound()) to behave differently depending on the object calling it.
Each subclass provides its own implementation of the method.
'''
# Base class
class animal:
  def sound(self):
    print("Animal Sound")
# Derived classes
class dog(animal):
  def sound(self):
    print("Bark")
class cat(animal):
  def sound(self):
    print("Meow")
class cow(animal):
  def sound(self):
    print("Moo Moo!")
# Demonstrating polymorphism
animals=[dog(),cat(),cow()]
for animal in animals:
  animal.sound()   # Same method name, different behavior
```

```python
# Explain abstraction in Python using the abc module.
# Requirements are Import ABC and abstractmethod from abc,Create an abstract class named Shape, Declare an abstract method area().
# Create Circle and Rectangle classes that implement area(),Demonstrate why an abstract class cannot be instantiated directly.
# Explain abstraction with a real-world example,Provide executable Python 3 code and sample output.

'''
🔹 Abstraction Concept
Abstraction means hiding implementation details and showing only the essential features.
In Python, abstraction is achieved using abstract classes and abstract methods.
Abstract classes act as blueprints for other classes.

Using the abc Module
ABC → Base class for defining abstract classes.
@abstractmethod → Decorator to mark methods that must be implemented in child classes.
Abstract classes cannot be instantiated directly.
'''
from abc import ABC,abstractmethod
# Abstract class
class shape(ABC):
  @abstractmethod
  def area(self):
    pass   #Abstract method, no implementation here
class Circle(shape):
  def __init__(self,radius):
    self.radius=radius
  def area(self):
    return 3.14*self.radius**2
class Rectangle(shape):
  def __init__(self,length,width):
    self.length=length
    self.width=width
  def area(self):
    return self.length*self.width
circle=Circle(5)
rectangle=Rectangle(4,6)
print("Circle Area:", circle.area())
print("Rectangle Area:", rectangle.area())
```

```python
# Write a Python program demonstrating method overriding.
# Requirements are Create a parent class named Vehicle with a start() method, Create Car and Bike subclasses, Override the start() method in both subclasses.
# Create objects and call start(),Explain how overriding supports runtime polymorphism.

'''
Method Overriding Concept---
Method overriding happens when a child class provides its own implementation of a method already defined in the parent class.
At runtime, Python decides which version of the method to call based on the object type.
This supports runtime polymorphism — the same method name behaves differently depending on the object.
'''
# Parent class
class Vehicle:
    def start(self):
        print("Vehicle is starting...")

# Child class Car overriding start()
class Car(Vehicle):
    def start(self):
        print("Car starts with a key ignition.")

# Child class Bike overriding start()
class Bike(Vehicle):
    def start(self):
        print("Bike starts with a kick or button.")

# Demonstration
v = Vehicle()
c = Car()
b = Bike()

v.start()   # Calls Vehicle's start()
c.start()   # Calls Car's overridden start()
b.start()   # Calls Bike's overridden start()

```

```python
# Explain method overloading in Python.
# Requirements are Explain why Python does not support traditional method overloading by defining multiple methods with the same name in one class.
# Demonstrate a flexible method using default arguments, Demonstrate an alternative using *args, Show how different numbers of arguments can be handled.
# Compare Python's approach with traditional method overloading in languages such as Java, Provide executable Python code and sample output.


# Why Python Does Not Support Traditional Method Overloading
# In languages like Java, you can define multiple methods with the same name but different parameter lists (e.g., add(int, int) and add(double, double)).
# In Python, if you define multiple methods with the same name in one class, the last definition overrides the previous ones.
# This is because Python is dynamically typed and does not check parameter types at compile time.

# 🔹 Flexible Alternatives in Python
# 1 Using Default Arguments python
class Calculator:
    def add(self, a=0, b=0, c=0):
        return a + b + c

calc = Calculator()
print("Add two numbers:", calc.add(5, 3))       # 8
print("Add three numbers:", calc.add(5, 3, 2))  # 10
print("Add with defaults:", calc.add())         # 0
#Default values allow flexibility without multiple definitions.

#2. Using *args (Variable Arguments) python
class Calculator:
    def add(self, *args):
        return sum(args)

calc = Calculator()
print("Add two numbers:", calc.add(5, 3))          # 8
print("Add three numbers:", calc.add(5, 3, 2))     # 10
print("Add five numbers:", calc.add(1, 2, 3, 4, 5)) # 15
#*args lets you pass any number of arguments, handled as a tuple.

#3. Handling Different Numbers of Arguments python
class Calculator:
    def multiply(self, *args):
        if len(args) == 2:
            return args[0] * args[1]
        elif len(args) == 3:
            return args[0] * args[1] * args[2]
        else:
            return "Unsupported number of arguments"

calc = Calculator()
print("Multiply two numbers:", calc.multiply(2, 3))       # 6
print("Multiply three numbers:", calc.multiply(2, 3, 4))  # 24
print("Multiply four numbers:", calc.multiply(2, 3, 4, 5)) # Unsupported
```

```python
# Composition Concept---
# Composition models a “has-a” relationship.
# Example: A Car has an Engine.
# Instead of inheriting from Engine, the Car class contains an Engine object.
# This allows flexible design where classes are built from smaller, reusable components.

# Engine class
class Engine:
    def start(self):
        print("Engine starts... Vroom!")
# Car class using composition
class Car:
    def __init__(self):
        # Car has an Engine
        self.engine = Engine()
    def start(self):
        print("Car is ready to start.")
        # Delegating work to Engine
        self.engine.start()

# Demonstration
my_car = Car()
my_car.start()
```
