```python
# Teach SQL window functions using the employees table.
# Write queries to:----
# 1. Rank employees by salary.
# 2. Compare RANK and DENSE_RANK.
# 3. Assign ROW_NUMBER within each department.
# 4. Find the highest-paid employee in each department.
# 5. Find the second-highest-paid employee in each department.
# 6. Calculate department-wise average salary without collapsing rows.
# 7. Calculate a running total of salaries.
# 8. Explain PARTITION BY and ORDER BY inside OVER().
# Provide expected output and note database compatibility.

import sqlite3
import pandas as pd

# Create in-memory SQLite DB
conn = sqlite3.connect(":memory:")
cur = conn.cursor()

# Create employees table
cur.execute("""CREATE TABLE employees (
    employee_id INTEGER PRIMARY KEY,
    employee_name TEXT NOT NULL,
    department_id INTEGER,
    salary INTEGER,
    hire_date DATE,
    city TEXT
)""")

# Insert sample employees
employees = [
(101,'Rahul',2,60000,'2020-05-10','Hyderabad'),
(102,'Sneha',1,45000,'2019-03-15','Delhi'),
(103,'Arjun',2,60000,'2021-07-20','Bangalore'),
(104,'Priya',3,70000,'2018-11-01','Mumbai'),
(105,'Kiran',4,50000,'2022-01-12','Chennai'),
(106,'Meena',1,45000,'2020-09-25','Delhi'),
(107,'Vikram',3,70000,'2017-06-30','Hyderabad'),
(108,'Anita',4,55000,'2021-12-05','Bangalore'),
(109,'Ravi',2,60000,'2019-08-18','Hyderabad'),
(110,'Divya',3,80000,'2022-04-10','Mumbai')
]
cur.executemany("INSERT INTO employees VALUES (?,?,?,?,?,?)", employees)

# Define window function queries
queries = [
("Rank employees by salary",
 """SELECT employee_name, salary,
           RANK() OVER (ORDER BY salary DESC) AS salary_rank
    FROM employees"""),

("Compare RANK vs DENSE_RANK",
 """SELECT employee_name, salary,
           RANK() OVER (ORDER BY salary DESC) AS rank_val,
           DENSE_RANK() OVER (ORDER BY salary DESC) AS dense_rank_val
    FROM employees"""),

("Assign ROW_NUMBER within each department",
 """SELECT employee_name, department_id, salary,
           ROW_NUMBER() OVER (PARTITION BY department_id ORDER BY salary DESC) AS row_num
    FROM employees"""),

("Highest-paid employee in each department",
 """SELECT employee_name, department_id, salary
    FROM (
        SELECT employee_name, department_id, salary,
               RANK() OVER (PARTITION BY department_id ORDER BY salary DESC) AS rnk
        FROM employees
    ) sub
    WHERE rnk = 1"""),

("Second-highest-paid employee in each department",
 """SELECT employee_name, department_id, salary
    FROM (
        SELECT employee_name, department_id, salary,
               RANK() OVER (PARTITION BY department_id ORDER BY salary DESC) AS rnk
        FROM employees
    ) sub
    WHERE rnk = 2"""),

("Department-wise average salary without collapsing rows",
 """SELECT employee_name, department_id, salary,
           AVG(salary) OVER (PARTITION BY department_id) AS dept_avg_salary
    FROM employees"""),

("Running total of salaries by hire_date",
 """SELECT employee_name, salary, hire_date,
           SUM(salary) OVER (ORDER BY hire_date) AS running_total
    FROM employees"""),

("Explain PARTITION BY and ORDER BY",
 """SELECT employee_name, department_id, salary,
           RANK() OVER (PARTITION BY department_id ORDER BY salary DESC) AS dept_rank
    FROM employees""")
]

# Execute and display results
for desc, q in queries:
    print(f"\n--- {desc} ---")
    df = pd.read_sql_query(q, conn)
    print(df)
```
