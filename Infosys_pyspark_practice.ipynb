PySpark is the Python interface for Apache Spark, used to process and analyze large-scale datasets across multiple machines. It lets you write Python code that Spark executes in a distributed way, so you can handle terabytes of data efficiently.

🔹 Real-Life Example (Topic-Based)
Imagine you’re working with an employees dataset (like the one we’ve been using):

Problem: A company has millions of employee records stored across different servers. HR wants to quickly find:

The highest-paid employee in each department

The average salary per city

Employees who earn above the company average



```python
# Install pyspark if not already installed
!pip install -q pyspark

from pyspark.sql import SparkSession

# Initialize SparkSession
spark = SparkSession.builder \
    .appName("EmployeeAnalysis") \
    .getOrCreate()

# Note: You will need to create and register the 'employees' temporary view
# with your actual data before running these queries, for example:
# df = spark.read.csv("path_to_employees.csv", header=True, inferSchema=True)
# df.createOrReplaceTempView("employees")

# Highest salary per department
try:
    spark.sql("""
    SELECT department_id, MAX(salary) AS max_salary
    FROM employees
    GROUP BY department_id
    """).show()
except Exception as e:
    print("Could not run query. Ensure the 'employees' view is registered.", e)

# Average salary per city
try:
    spark.sql("""
    SELECT city, AVG(salary) AS avg_salary
    FROM employees
    GROUP BY city
    """).show()
except Exception as e:
    pass

# Employees earning above company average
try:
    spark.sql("""
    SELECT employee_name, salary
    FROM employees
    WHERE salary > (SELECT AVG(salary) FROM employees)
    """).show()
except Exception as e:
    pass
```

```python
# Install PySpark in Colab
!pip install pyspark

# Import SparkSession
from pyspark.sql import SparkSession
from pyspark.sql import Row

# Create SparkSession
spark = SparkSession.builder \
    .appName("PySpark Small Dataset Example") \
    .getOrCreate()

# Create a small employees dataset
employees_data = [
    Row(employee_id=101, employee_name="Rahul", department_id=2, salary=60000, city="Hyderabad"),
    Row(employee_id=102, employee_name="Sneha", department_id=1, salary=45000, city="Delhi"),
    Row(employee_id=103, employee_name="Arjun", department_id=2, salary=60000, city="Bangalore"),
    Row(employee_id=104, employee_name="Priya", department_id=3, salary=70000, city="Mumbai"),
    Row(employee_id=105, employee_name="Kiran", department_id=4, salary=50000, city="Chennai")
]

# Create DataFrame
employees_df = spark.createDataFrame(employees_data)

# Show DataFrame
employees_df.show()
```
